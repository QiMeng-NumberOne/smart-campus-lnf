```python
import cv2
from ultralytics import YOLO

def start_campus_detection():
    # 1. 加载你亲自训练出来的模型！
    # 路径指向你 runs 文件夹里效果最好的那个权重文件
    # 建议你把 runs/detect/campus_lost_v84/weights/best.pt 复制到项目根目录，方便调用
    model_path = 'runs/detect/train2/weights/best.pt' 
    
    try:
        model = YOLO(model_path)
        print(f"✅ 成功加载专属模型: {model_path}")
    except Exception as e:
        print(f"❌ 加载模型失败，请检查路径是否正确！错误信息: {e}")
        return

    # 2. 打开摄像头 (0 通常是内置摄像头)
    cap = cv2.VideoCapture(0)
    
    if not cap.isOpened():
        print("❌ 无法打开摄像头")
        return

    print("🚀 实时识别已开启！把钥匙、校园卡或眼镜放到摄像头前试试吧~")
    print("👉 按下键盘上的 'q' 键退出")

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        # 3. 使用你的模型进行推理
        # conf=0.5 表示只显示置信度大于 50% 的结果
        results = model(frame, conf=0.5, stream=True)

        # 4. 绘制识别结果
        for r in results:
            annotated_frame = r.plot() # 这行会自动画出你 data.yaml 里定义的标签名

        # 5. 显示画面
        cv2.imshow("Campus Lost & Found - Realtime AI", annotated_frame)

        # 按 'q' 退出
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()

if __name__ == '__main__':
    start_campus_detection()
```
